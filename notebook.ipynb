## 1. Why version control matters

So far, every notebook in this course has lived inside a **git repository** — you just haven't needed to think about it. Git tracks every change to every file over time, which means you can:

- See exactly what changed, when, and why
- Undo a mistake by going back to an earlier version
- Work on a feature without breaking the working version (using branches)
- Collaborate with others without overwriting each other's work

## 3. The core workflow: add, commit, push, pull

The everyday git cycle has four commands:

| Command | What it does |
|---|---|
| `git add <file>` | Stage a file — mark it to be included in the next commit |
| `git commit -m "message"` | Save a snapshot of all staged changes, with a message describing *why* |
| `git push` | Upload your commits to the remote repository (e.g. GitHub) |
| `git pull` | Download and merge any new commits from the remote |

A typical session looks like this:

```bash
git pull                          # get the latest changes first
# ... edit some files ...
git status                        # see what changed
git add notebook.ipynb            # stage the file(s) you want to commit
git commit -m "Complete Unit 6 exercises"
git push                          # share your commit
```

**Good commit messages** describe *why* a change was made, not just *what* — e.g. `"Fix off-by-one error in transaction loop"` is more useful than `"update code"`.

## 5. PEP 8 — Python's style guide

**PEP 8** is the official style guide for Python code. Following it makes your code easier for *anyone* (including future you) to read. A few of the most common rules:

- Use `snake_case` for variable and function names (not `camelCase`)
- Put spaces around operators: `a = b + c`, not `a=b+c`
- Use 4 spaces per indentation level (never tabs)
- Limit lines to 79 characters where practical
- Two blank lines between top-level function definitions

### Before: hard to read


````python
def calcTotal(Items,gstRate=0.09):
  total=0
  for i in Items:
      total=total+i['price']*(1+gstRate)
  return total
items=[{'price':100},{'price':250}]
print(calcTotal(items))
````

### After: PEP 8 compliant


````python
def calculate_total(items, gst_rate=0.09):
    total = 0
    for item in items:
        total = total + item["price"] * (1 + gst_rate)
    return total


items = [{"price": 100}, {"price": 250}]
print(calculate_total(items))
````

Both versions produce the same result — but the second one is far easier to read, debug, and extend. Your devcontainer is already configured with the **Black** formatter (`editor.formatOnSave: true`), which automatically reformats your code towards PEP 8 every time you save a file.


## 6. A first look at pandas

Throughout this course, you've used the **list-of-dicts** pattern to represent tabular data, and written loops and comprehensions to filter, transform, and aggregate it (Units 2 and 4). The `pandas` library — already included in your devcontainer — gives you a `DataFrame`: a table-like structure built for exactly this kind of work, with most of those operations built in.


````python
import pandas as pd

sales = [
    {"date": "2026-02-02", "item": "Coffee", "quantity": 3, "unit_price": 4.50},
    {"date": "2026-02-02", "item": "Tea", "quantity": 1, "unit_price": 3.50},
    {"date": "2026-02-03", "item": "Coffee", "quantity": 5, "unit_price": 4.50},
    {"date": "2026-02-03", "item": "Muffin", "quantity": 4, "unit_price": 3.00},
    {"date": "2026-02-04", "item": "Coffee", "quantity": 2, "unit_price": 4.50},
]

df = pd.DataFrame(sales)
df
````

````python
# Transform: a new column, calculated for every row at once
df["line_total"] = df["quantity"] * df["unit_price"]
df
````

````python
# Filter: rows where line_total is over $15
df[df["line_total"] > 15]
````

````python
# Aggregate: total revenue per item
df.groupby("item")["line_total"].sum()
````

Compare this to Unit 5's pipeline: reading the CSV, converting types, filtering large transactions, and totalling revenue per item took roughly 20 lines of loops and dicts. Here it's four lines — `pd.DataFrame(...)`, a column assignment, a boolean filter, and a `groupby`. Pandas doesn't replace what you've learned; it's *built on* the same ideas (rows as records, columns as fields) and automates the loops you'd otherwise write by hand.
